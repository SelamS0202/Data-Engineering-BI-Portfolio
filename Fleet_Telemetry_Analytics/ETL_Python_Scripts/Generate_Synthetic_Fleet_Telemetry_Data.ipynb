```python
import random
import pandas as pd
from datetime import datetime, timedelta 
```

#### Dim_vehicle

```python
# 50 vehicles labled V001 to V050
vehicle_ids = ["V"+ str(i).zfill(3) for i in range(1,51)]
vehicle_ids [:5]
```

```python
# assign vehicle types
vehicle_type = [random.choice(['Truck', 'Van','Refrigerated']) for _ in range (50)]
vehicle_type[:5]
```

```python
# assign regions
regions = [random.choice(['North','South','East','West']) for _ in range(50)]
regions[: 6]
```

```python
# assign fuel capacity
fuel_capacity = [random.randint(60,300) for _ in range(50)]
fuel_capacity[:7]
```

```python
# combine data
dim_vehicle = pd.DataFrame({
    'vehicle_id': vehicle_ids,
    'vehicle_type': vehicle_type,
    'region': regions,
    'fuel_capacity': fuel_capacity})
dim_vehicle.head(10)
```

```python
# save as csv
dim_vehicle.to_csv('dim_vehicle.csv', index= False)
```

#### Telemetry Data

```python
num_vehicles = 50
num_days = 30
records_per_day = 100
start_date = datetime(2026, 1,1)
```

```python
#  empty list for data
data = []
```

```python
for vehicle in vehicle_ids:
    for day in range(num_days):
        current_date = start_date + timedelta(days = day) # increment day by 1
        for record in range(records_per_day):
            timestamp = current_date + timedelta(minutes = record * 10) # each record is 10 minutes apart
            speed = random.randint(0,12) # random speed between  0 -120 km/h
            fuel_level = random.randint(10,100)
            distance = round (speed/60,2)
            engine_temp = random.randint(70, 110)
            data.append([timestamp, vehicle, speed, fuel_level, distance, engine_temp]) #add row to Data
col = ['timestamp', 'vehicle_id', 'speek_kmh','fuel_level_%', 'distance_km', 'engine_temp']
df = pd.DataFrame(data, columns = col)
df.head(5)
```

```python
# save to csv
df.to_csv('raw_telematry.csv', index = False)
```
